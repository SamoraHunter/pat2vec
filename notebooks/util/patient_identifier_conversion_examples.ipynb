# Patient Identifier Conversion

This notebook demonstrates how to convert between patient identifier types using the `epic_patients` index. The index contains records where each record has all identifiers (Hospital Number, Durable Key, MRN, NHS Number) linked together.

## Available Identifiers in epic_patients Index
- `patient_HospitalNumber` - Hospital number(s) (can be multiple, comma-separated)
- `patient_DurableKey` - Unique durable key for the patient
- `Patient_MRN` - Medical Record Number
- `patient_NhsNumber` - NHS Number

## Setup

Use `testing=True` in config to generate dummy data instead of connecting to Elasticsearch.


```python
#!rm -r patient_identifier_conversion_test

from pat2vec.util.config_pat2vec import config_class
from pat2vec.main_pat2vec import main

config_obj = config_class(
    testing=True,
    verbosity=9,
    proj_name="patient_identifier_conversion_test",
)

pat2vec_obj = main(config_obj=config_obj, cogstack=True)
```

## Conversion Functions


```python
from pat2vec.util.patient_identifier_conversion import (
    convert_hospital_numbers_to_durable_keys,
    convert_durable_keys_to_hospital_numbers,
    convert_nhs_numbers_to_durable_keys,
    convert_durable_keys_to_nhs_numbers,
    convert_mrns_to_durable_keys,
    convert_durable_keys_to_mrns,
)
```

## Conversion Examples

### 1. NHS Numbers → Durable Keys


```python
input_nhs_numbers = ["1234567890", "0987654321"]
durable_keys, missing = convert_nhs_numbers_to_durable_keys(
    input_nhs_numbers, pat2vec_obj
)

print(f"Input NHS numbers: {input_nhs_numbers}")
print(f"Durable keys found: {len(durable_keys)}")
```

### 2. MRNs → Durable Keys


```python
input_mrns = ["ABC123", "XYZ789"]
durable_keys, missing = convert_mrns_to_durable_keys(input_mrns, pat2vec_obj)

print(f"Input MRNs: {input_mrns}")
print(f"Durable keys found: {len(durable_keys)}")
```

### 3. Hospital Numbers → Durable Keys


```python
input_hospital_numbers = ["123456", "789012"]
durable_keys, missing = convert_hospital_numbers_to_durable_keys(
    input_hospital_numbers, pat2vec_obj
)

print(f"Input hospital numbers: {input_hospital_numbers}")
print(f"Durable keys found: {len(durable_keys)}")
```

### 4. Durable Keys → Hospital Numbers


```python
input_durable_keys = ["DURABLE001", "DURABLE002"]
hospital_numbers, missing_hn = convert_durable_keys_to_hospital_numbers(
    input_durable_keys, pat2vec_obj
)

print(f"Input durable keys: {input_durable_keys}")
print(f"Hospital numbers found: {len(hospital_numbers)}")
```

### 5. Durable Keys → NHS Numbers


```python
input_durable_keys = ["DURABLE001", "DURABLE002"]
nhs_numbers, missing_nhs = convert_durable_keys_to_nhs_numbers(
    input_durable_keys, pat2vec_obj
)

print(f"Input durable keys: {input_durable_keys}")
print(f"NHS numbers found: {len(nhs_numbers)}")
```

### 6. Durable Keys → MRNs


```python
input_durable_keys = ["DURABLE001", "DURABLE002"]
mrns, missing_mrn = convert_durable_keys_to_mrns(input_durable_keys, pat2vec_obj)

print(f"Input durable keys: {input_durable_keys}")
print(f"MRNs found: {len(mrns)}")
```

### 7. Complete Conversion Chain: NHS Numbers → MRNs


```python
# Step 1: NHS numbers to durable keys
input_nhs_numbers = ["1234567890", "0987654321"]
durable_keys, missing_nhs = convert_nhs_numbers_to_durable_keys(
    input_nhs_numbers, pat2vec_obj
)

# Step 2: Durable keys to MRNs (pass all keys, not just first one)
output_mrns, missing_mrn = convert_durable_keys_to_mrns(durable_keys, pat2vec_obj)

print(f"Input NHS numbers: {input_nhs_numbers}")
print(f"Durable keys found: {len(durable_keys)}")
print(f"MRNs found: {len(output_mrns)}")
```

## Summary

The conversion functions work by:
1. Querying the `epic_patients` index with a list of source identifiers
2. Filtering results to return only the desired target identifier field
